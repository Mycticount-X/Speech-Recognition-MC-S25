## Predictor / Transcriber

```python
def Transcriber (audiopath):
    audio, sr = torchaudio.load(audiopath)

    if sr != bundle.sample_rate:
        audio = torchaudio.functional.resample(audio, sr, bundle.sample_rate)

    with torch.inference_mode():
        emission, _ = model(audio)

    class GreedyCTCDecoder (torch.nn.Module):
        def __init__(self, labels):
            super().__init__()
            self.labels = labels
            self.blank = 0

        def forward(self, emission: torch.Tensor):
            idx = torch.argmax(emission, dim=-1)
            idx = torch.unique_consecutive(idx, dim=-1)
            idx = [i for i in idx if i != self.blank]
            return "".join([self.labels[i] for i in idx])

    decoder = GreedyCTCDecoder(bundle.get_labels())
    text = decoder(emission[0])
    text = text.replace("|", " ")
    return text
```

## Support Function

```python
def ExportMenu (audio: AudioSegment):
    filename = None

    while True:
        print('Audio File (.wav): ')
        filename = input()

        if filename.endswith('.wav'):
            break
        else:
            print("Invalid Audio Name")

    audio.export(filename, "wav")

def adj_pitch (audio: AudioSegment, pitch):
    buffer = io.BytesIO()
    audio.export(buffer, "wav")
    buffer.seek(0)

    signal, sr = librosa.load(buffer, mono=False)
    signal = librosa.effects.pitch_shift(signal, sr=sr, n_steps=pitch)

    buffer = io.BytesIO()
    sf.write(buffer, signal, samplerate=sr, format="wav")
    buffer.seek(0)

    return AudioSegment.from_wav(buffer)

def add_whitenoise (audio: AudioSegment, rate):
    buffer = io.BytesIO()
    audio.export(buffer, "wav")
    buffer.seek(0)

    signal, sr = librosa.load(buffer, mono=False)
    noice = np.random.normal(0, signal.std(), signal.shape)
    signal = signal + noice * rate

    buffer = io.BytesIO()
    sf.write(buffer, signal, samplerate=sr, format="wav")
    buffer.seek(0)

    return AudioSegment.from_wav(buffer)

def add_repeat (audio: AudioSegment, repeat):
    # Note: Sesuai namanya, repeat / mengulang
    # Hasil = original audio + jumlah repeat (r = 2, hasil = 3)
    return audio * (repeat + 1)

def add_fadein (audio: AudioSegment, dur):
    return audio.fade_in(dur)
```
